```python
import os

import requests

RAG_URL="http://localhost:8001/internal/query"
INTERNAL_HEADERS = {"X-Internal-Secret": os.environ.get("INTERNAL_SERVICE_SECRET", "")}

def run_rag(question):
    response= requests.post(
        RAG_URL,
        json={"question":question},
        headers=INTERNAL_HEADERS,
        timeout=120
    )
    response.raise_for_status()

    return response.json()
```

```python
answer= run_rag("what is the name of the insurance policy?")
print(answer)
```

```python
results = []

for i, example in enumerate(
    client.list_examples(dataset_id=dataset.id), 
    start=1
    ):

    question = example.inputs["question"]

    try:
        rag_response= run_rag(question)

        results.append({
            "question": question,
            "expected_answer": example.outputs["expected_answer"],
            "actual_answer": rag_response["answer"],
            "context": rag_response["context"],
            "status": "success"
        })

        print(f"[{i}/35] ✓ {question}")

    except Exception as e:
        results.append({
            "question": question,
            "expected_answer": example.outputs["expected_answer"],
            "actual_answer": None,
            "context": [],
            "status": f"error: {e}"
        })

        print(f"[{i}/35] ✗ {question}")
```

## LLM-based Answer Correctness evaluator

```python
from langchain_anthropic import ChatAnthropic

judge_llm = ChatAnthropic(
    model="claude-sonnet-4-6",
    temperature=0
)

def answer_correctness(run, example):
    # Handle 'run' whether it's a LangSmith Run object or a dictionary
    if isinstance(run, dict):
        actual_answer = run.get("outputs", {}).get("answer", "")
    else:
        actual_answer = getattr(run, "outputs", {}).get("answer", "")

    # Handle 'example' whether it's a LangSmith Example object or a dictionary
    if isinstance(example, dict):
        expected_answer = example.get("outputs", {}).get("expected_answer", "")
    else:
        expected_answer = example.outputs.get("expected_answer", "")

    prompt = f"""
You are evaluating the answer produced by a RAG system.

Compare the ACTUAL ANSWER against the EXPECTED ANSWER.

EXPECTED ANSWER:
{expected_answer}

ACTUAL ANSWER:
{actual_answer}

Determine whether the actual answer is factually consistent with
the expected answer.

Important:
- Different wording is acceptable.
- Additional correct explanation is acceptable.
- Do not require exact wording.
- If the actual answer contradicts the expected answer, mark it incorrect.
- If the actual answer contains the expected information but is more detailed,
  mark it correct.
- For "not available in the provided documents" questions, the answer should
  correctly indicate that the requested information is unavailable.

Return ONLY one of these:
CORRECT
INCORRECT
"""

    response = judge_llm.invoke(prompt)
    verdict = response.content.strip().upper()
    score = 1 if verdict == "CORRECT" else 0

    return {
        "key": "answer_correctness",
        "score": score,
        "comment": verdict
    }
```

```python
evaluation_results = []

for i, result in enumerate(results, start=1):

    test_run = {
        "outputs": {
            "answer": result["actual_answer"]
        }
    }

    test_example = {
        "outputs": {
            "expected_answer": result["expected_answer"]
        }
    }

    evaluation = answer_correctness(
        test_run,
        test_example
    )

    evaluation_results.append({
        "question": result["question"],
        "expected_answer": result["expected_answer"],
        "actual_answer": result["actual_answer"],
        "score": evaluation["score"],
        "comment": evaluation["comment"]
    })

    print(
        f"[{i}/35] "
        f"{evaluation['comment']} "
        f"→ {result['question']}"
    )

print("\nEvaluation completed!")
print("Total evaluated:", len(evaluation_results))
```

## Faithfulness evaluator

```python
def faithfulness(run, example):
    actual_answer = run.get("outputs", {}).get("answer", "")
    context = run.get("outputs", {}).get("context", [])

    context_text = "\n\n--- DOCUMENT ---\n\n".join(context)

    prompt = f"""
You are evaluating the faithfulness of a RAG system's answer.

QUESTION:
{example.get("inputs", {}).get("question", "")}

RETRIEVED CONTEXT:
{context_text}

ANSWER:
{actual_answer}

Determine whether the ANSWER is fully supported by the RETRIEVED CONTEXT.

Rules:
- The answer must be supported by the retrieved context.
- Do not use outside knowledge.
- If the answer contains claims that are not supported by the context,
  mark it INCORRECT.
- If the answer is fully supported by the context, mark it CORRECT.
- Different wording is acceptable.
- For an answer saying information is unavailable, check whether the
  retrieved context actually supports that conclusion.

Return ONLY:
CORRECT
or
INCORRECT
"""

    response = judge_llm.invoke(prompt)

    verdict = response.content.strip().upper()

    return {
        "key": "faithfulness",
        "score": 1 if verdict == "CORRECT" else 0,
        "comment": verdict
    }
```

```python
test_result = results[0]

test_run = {
    "outputs": {
        "answer": test_result["actual_answer"],
        "context": test_result["context"]
    }
}

test_example = {
    "inputs": {
        "question": test_result["question"]
    }
}

faithfulness_result = faithfulness(
    test_run,
    test_example
)

print(faithfulness_result)
```

```python
faithfulness_results = []

for i, result in enumerate(results, start=1):

    test_run = {
        "outputs": {
            "answer": result["actual_answer"],
            "context": result["context"]
        }
    }

    test_example = {
        "inputs": {
            "question": result["question"]
        }
    }

    evaluation = faithfulness(
        test_run,
        test_example
    )

    faithfulness_results.append({
        "question": result["question"],
        "score": evaluation["score"],
        "comment": evaluation["comment"]
    })

    print(
        f"[{i}/35] "
        f"{evaluation['comment']} "
        f"→ {result['question']}"
    )

print("\nFaithfulness evaluation completed!")
print("Total evaluated:", len(faithfulness_results))
```

## Answer Relevance

```python
def answer_relevance(run, example):
    actual_answer = run.get("outputs", {}).get("answer", "")
    question = example.get("inputs", {}).get("question", "")

    prompt = f"""
You are evaluating the relevance of a RAG system's answer.

QUESTION:
{question}

ANSWER:
{actual_answer}

Determine whether the answer directly addresses the user's question.

Rules:
- The answer should directly address what was asked.
- Do not judge factual correctness here.
- Focus only on whether the answer is relevant to the question.
- Concise answers are acceptable.
- Additional relevant explanation is acceptable.
- If the answer discusses unrelated information, mark it INCORRECT.
- For questions whose information is unavailable in the documents,
  an appropriate response stating that the information is unavailable
  is considered relevant.

Return ONLY:
CORRECT
or
INCORRECT
"""

    response = judge_llm.invoke(prompt)

    verdict = response.content.strip().upper()

    return {
        "key": "answer_relevance",
        "score": 1 if verdict == "CORRECT" else 0,
        "comment": verdict
    }
```

```python
test_result = results[0]

test_run = {
    "outputs": {
        "answer": test_result["actual_answer"]
    }
}

test_example = {
    "inputs": {
        "question": test_result["question"]
    }
}

relevance_result = answer_relevance(
    test_run,
    test_example
)

print(relevance_result)
```

```python
relevance_results = []

for i, result in enumerate(results, start=1):

    test_run = {
        "outputs": {
            "answer": result["actual_answer"]
        }
    }

    test_example = {
        "inputs": {
            "question": result["question"]
        }
    }

    evaluation = answer_relevance(
        test_run,
        test_example
    )

    relevance_results.append({
        "question": result["question"],
        "score": evaluation["score"],
        "comment": evaluation["comment"]
    })

    print(
        f"[{i}/35] "
        f"{evaluation['comment']} "
        f"→ {result['question']}"
    )

print("\nAnswer Relevance evaluation completed!")
print("Total evaluated:", len(relevance_results))
```

```python
total = len(relevance_results)

relevant = sum(
    1 for r in relevance_results
    if r["score"] == 1
)

relevance_score = (relevant / total) * 100

print(f"Total questions : {total}")
print(f"Relevant        : {relevant}")
print(f"Not relevant    : {total - relevant}")
print(f"Answer Relevance: {relevance_score:.2f}%")
```

## Context/Retrieval evaluation

```python
def retrieval_quality(run, example):
    question = example.get("inputs", {}).get("question", "")
    context = run.get("outputs", {}).get("context", [])

    context_text = "\n\n--- DOCUMENT ---\n\n".join(context)

    prompt = f"""
You are evaluating the retrieval quality of a RAG system.

QUESTION:
{question}

RETRIEVED DOCUMENTS:
{context_text}

Determine whether the retrieved documents contain the information
necessary to answer the question.

Rules:
- If the retrieved context contains sufficient information to answer
  the question, return CORRECT.
- If important information needed to answer the question is missing,
  return INCORRECT.
- The documents do not need to contain the exact wording of the answer.
- Semantic equivalence is acceptable.
- If the question is out-of-context/unanswerable from the provided documents, 
  retrieval is considered CORRECT if the system appropriately scanned the correct documents 
  and found no matching data (i.e., it should not be penalized for missing information 
  that does not exist in the source material).

Return ONLY:
CORRECT
or
INCORRECT
"""

    response = judge_llm.invoke(prompt)

    verdict = response.content.strip().upper()

    return {
        "key": "retrieval_quality",
        "score": 1 if verdict == "CORRECT" else 0,
        "comment": verdict
    }
```

```python
test_result = results[0]

test_run = {
    "outputs": {
        "context": test_result["context"]
    }
}

test_example = {
    "inputs": {
        "question": test_result["question"]
    }
}

retrieval_result = retrieval_quality(
    test_run,
    test_example
)

print(retrieval_result)
```

```python
retrieval_results = []

for i, result in enumerate(results, start=1):

    test_run = {
        "outputs": {
            "context": result["context"]
        }
    }

    test_example = {
        "inputs": {
            "question": result["question"]
        }
    }

    evaluation = retrieval_quality(
        test_run,
        test_example
    )

    retrieval_results.append({
        "question": result["question"],
        "score": evaluation["score"],
        "comment": evaluation["comment"]
    })

    print(
        f"[{i}/35] "
        f"{evaluation['comment']} "
        f"→ {result['question']}"
    )

print("\nRetrieval Quality evaluation completed!")
print("Total evaluated:", len(retrieval_results))
```

```python
total = len(retrieval_results)

retrieved_correctly = sum(
    1 for r in retrieval_results
    if r["score"] == 1
)

retrieval_score = (retrieved_correctly / total) * 100

print(f"Total questions      : {total}")
print(f"Retrieved correctly  : {retrieved_correctly}")
print(f"Retrieval failures   : {total - retrieved_correctly}")
print(f"Retrieval Quality    : {retrieval_score:.2f}%")
```
